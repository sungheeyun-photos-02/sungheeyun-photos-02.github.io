```python
import os
import shutil
```

```python
dirname = os.curdir
```

```python
filename_list = os.listdir(dirname)
```

```python
for filename in filename_list:
    filename_fullpath = os.path.join(dirname, filename)
    # print(filename_fullpath,os.path.isdir(filename_fullpath))
    if os.path.isdir(filename_fullpath):
        # print(os.listdir(filename_fullpath))
        file_list = [
            x for x in os.listdir(filename_fullpath) if os.path.splitext(x)[1].lower() == ".jpg"
        ]
        for file in file_list:
            from_file = os.path.join(filename_fullpath, file)
            to_file = os.path.join(os.curdir, file)
            print("cp %s %s" % (from_file, to_file))
            shutil.copy2(from_file, to_file)
```
